```python
import numpy as np

vocab = ["i", "am", "tired"]
word_to_index = {w: i for i, w in enumerate(vocab)}
index_to_word = {i: w for w, i in word_to_index.items()}

def one_hot_vector(idx, dim):
    vec = np.zeros(dim)
    vec[idx] = 1.0
    return vec

inputs = [one_hot_vector(word_to_index[w], len(vocab)) for w in ["i", "am"]]
target = one_hot_vector(word_to_index["tired"], len(vocab))

hidden_dim = 2
np.random.seed(42)
W_input_hidden = np.array([[0.45, -0.38],
                           [-0.25, 0.75],
                           [0.65, -0.55]])
W_hidden_hidden = np.array([[0.18, -0.12],
                            [0.48, 0.28]])
W_hidden_output = np.array([[0.55, -0.22, 0.33],
                            [-0.15, 0.68, 0.45]])

def softmax(vec):
    e = np.exp(vec - np.max(vec))
    return e / e.sum()

def mse_loss(pred, true):
    return np.mean((pred - true) ** 2)

h_prev = np.zeros(hidden_dim)

for x_t in inputs:
    h_in = W_input_hidden.T.dot(x_t) + W_hidden_hidden.T.dot(h_prev)
    h_prev = np.tanh(h_in)

logits = W_hidden_output.T.dot(h_prev)
probs = softmax(logits)

loss_value = mse_loss(probs, target)

dL_dy = 2 * (probs - target) / len(target)
dW_hidden_output = np.outer(h_prev, dL_dy)

print("Predicted probabilities:")
for i, p in enumerate(probs):
    print(f"  {index_to_word[i]}: {p:.4f}")

pred_idx = np.argmax(probs)
print(f"\nPredicted next word: '{index_to_word[pred_idx]}'")
print(f"Loss (MSE): {loss_value:.6f}")
print("\nGradient of W_hidden_output:")
print(dW_hidden_output)
```
